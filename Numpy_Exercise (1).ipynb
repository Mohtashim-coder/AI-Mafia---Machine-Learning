#### 15. Create a 2d array with 1 on the border and 0 inside

````python
def create_border_array(rows, cols):

    arr = np.zeros((rows, cols))
    arr[0, :] = 1 # top most row
    arr[-1, :] = 1 # bottom most row
    arr[:, 0] = 1 # left most column
    arr[:, -1] = 1 # right most column

    return arr

result = create_border_array(10, 10)
print(result)
````

#### 19. Create a 8x8 matrix and fill it with a checkerboard pattern

````python
# create 8x8 array of zeros
checkerboard = np.zeros((8, 8), dtype = int)

checkerboard[1::2, ::2] = 1 # odd rows, even columns
checkerboard[::2, 1::2] = 1 # even rows, odd columns

print(checkerboard)
````
